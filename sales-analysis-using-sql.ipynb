## Create SQL Environment

```python
import pandas as pd
df=pd.read_csv("/kaggle/input/sales-forecasting/train.csv")
df.columns
```

## Create SQL environment

```python
import sqlite3
conn=sqlite3.connect("superstore.db")
df.to_sql("sales",conn,index=False,if_exists="replace")
```

# SQL ANALYSIS QUESTIONS

## Q1. How many total orders are there?

```python
query='''
SELECT COUNT(DISTINCT "Order ID") AS total_orders
FROM sales
'''
pd.read_sql_query(query,conn)
```

## Q2. Total Sales?

```python
query='''
SELECT SUM(Sales) AS total_sales
FROM sales
'''
pd.read_sql_query(query,conn)
```

## Q3. Sales by category?

```python
query='''
SELECT Category, SUM(Sales) AS total_sales
FROM sales
GROUP BY Category
ORDER BY total_sales DESC
'''
pd.read_sql_query(query,conn)
```

## Q4. Top 5 sub-categories by sales?

```python
query='''
SELECT "Sub-Category", SUM(Sales) AS total_sales
FROM sales
GROUP BY "Sub-Category"
ORDER BY total_sales DESC
LIMIT 5
'''
pd.read_sql_query(query,conn)
```

## Q5. Sales by region?

```python
query='''
SELECT Region, SUM(Sales) as total_sales
FROM sales
GROUP BY Region
ORDER BY total_sales DESC
'''
pd.read_sql_query(query,conn)
```

## Q6. Average sales per order

```python
query='''
SELECT AVG(Sales) AS avg_sales
FROM sales
'''
pd.read_sql_query(query,conn)
```

## Q7. Top 5 customers by sales?

```python
query='''
SELECT "Customer Name", SUM(Sales) AS total_sales
FROM sales
GROUP BY "Customer Name"
ORDER BY total_sales DESC
LIMIT 5
'''
pd.read_sql_query(query,conn)
```

## Q8. Monthly sales trend?

```python
query='''
SELECT
    substr("Order Date",1,7) AS month,
    SUM(Sales) AS monthly_sales
FROM sales
GROUP BY month
ORDER BY month
'''
pd.read_sql_query(query,conn)
```

## Key Insights:

- Sales are concentrated in a few major categories and sub-categories.
- Certain regions contribute significantly more to overall revenue.
- A small group of customers drives a large portion of sales.
- Monthly sales trends suggest seasonal demand patterns.
